```python
import os
from pathlib import Path
from pypdf import PdfWriter

def Merge_PDF():
    """
    PDFファイルを研究室ごとに一つのファイルにする
    """
    # パス
    output_PDF_dir = '/content/drive/MyDrive/請求書発行/請求書/PDF'
    output_labo_dir = '/content/drive/MyDrive/請求書発行/請求書/LaboPDF'

    # 出力先フォルダが存在しない場合は作成
    os.makedirs(output_labo_dir, exist_ok=True)

    Read_folder_path = Path(output_PDF_dir)
    pdf_files = sorted(Read_folder_path.glob('*.pdf'))

    # 研究室ごとにファイルをグループ化する
    # ファイル名形式: 請求書_[研究室名]_[装置キー].pdf
    labo_pdf_map = {}
    for pdf_path in pdf_files:
        filename = pdf_path.name
        if filename.startswith('請求書_'):
            parts = filename.split('_')
            if len(parts) >= 2:
                labo_name = parts[1]
                if labo_name not in labo_pdf_map:
                    labo_pdf_map[labo_name] = []
                labo_pdf_map[labo_name].append(pdf_path)

    # 各研究室のPDFを結合して出力
    # items() を使うと(key, value)を取り出せる
    for labo_name, paths in labo_pdf_map.items():
        writer = PdfWriter()
        for path in paths:
            writer.append(path)

        output_file_path = Path(output_labo_dir) / f"請求書_{labo_name}.pdf"
        with open(output_file_path, "wb") as f_out: # 新規作成または既存ファイルの上書き
            writer.write(f_out)
        writer.close()
        print(f"結合完了: {output_file_path.name} ({len(paths)}件のPDFを結合)")
```

```python
def main():
  Merge_PDF()
```

```python
if __name__=="__main__":
  main()
```
